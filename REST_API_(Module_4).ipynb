```python
from flask import Flask, request, jsonify
from ocr.ocr_engine import extract_text
from ocr.text_validation import validate_aadhaar

app = Flask(__name__)

@app.route("/verify", methods=["POST"])
def verify():
    image = request.files["file"]
    image.save("temp.jpg")

    text = extract_text("temp.jpg")
    aadhaar = validate_aadhaar(text)

    return jsonify({
        "extracted_text": text,
        "aadhaar_status": aadhaar
    })

if __name__ == "__main__":
    app.run(debug=True)
```
